###Save the usable utterance list to one CSV & parse one MoCap .txt file into a clean NumPy array.

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
def build_usable_csv(data_dir: Path, shared_folder: Path) -> pd.DataFrame:
    root = next(data_dir.rglob("IEMOCAP_full_release"), data_dir)
    csv_path = next(data_dir.rglob("iemocap_full_dataset.csv"), None)
    df = pd.read_csv(csv_path)

    # the MoCap file sits at the same path as the wav, but in a
    # different folder and with a .txt extension
    def to_mocap_rel(wav_rel: str) -> str:
        return (wav_rel
                .replace("/sentences/wav/", "/sentences/MOCAP_rotated/")
                .replace(".wav", ".txt"))

    df["mocap_rel"] = df["path"].apply(to_mocap_rel)
    df["has_mocap"] = df["mocap_rel"].apply(lambda r: (root / r).exists())

    # 4-class benchmark labels (happy + excited are merged, as in the
    # standard IEMOCAP protocol used by published baselines)
    map4 = {"neu": "neutral", "sad": "sad",
            "ang": "angry", "hap": "happy", "exc": "happy"}
    df["emo4"] = df["emotion"].map(map4)

    usable = df[df["emo4"].notna() & df["has_mocap"]].copy()
    usable["utt_id"] = usable["path"].apply(lambda p: Path(p).stem)

    out = (usable[["utt_id", "session", "gender", "emo4",
                   "emotion", "agreement", "path", "mocap_rel"]]
           .rename(columns={"path": "wav_rel"})
           .reset_index(drop=True))

    save_path = shared_folder / "usable_utterances.csv"
    out.to_csv(save_path, index=False)
    print(f"Saved {len(out)} rows -> {save_path}")
    print(out["emo4"].value_counts())
    return out
```

```python
N_FACE_MARKERS = 53          # markers 01..53
N_HEAD_MARKERS = 2           # markers 60, 61 (dropped)


def load_mocap(txt_path) -> np.ndarray:
    """Return facial marker motion as an array of shape (frames, 53, 3).

    Missing markers are NaN. Frame# and Time columns are dropped.
    """
    lines = Path(txt_path).read_text(errors="ignore").splitlines()
    data_lines = [ln for ln in lines[2:] if ln.strip()]   # skip 2 headers

    rows = []
    for ln in data_lines:
        parts = ln.split()
        coords = [float(x) for x in parts[2:]]            # drop Frame#, Time
        rows.append(coords)

    arr = np.asarray(rows, dtype=np.float32)              # (frames, 165)

    # keep the first 53 markers (159 values); drop the 2 head markers
    arr = arr[:, : N_FACE_MARKERS * 3]                    # (frames, 159)
    arr = arr.reshape(arr.shape[0], N_FACE_MARKERS, 3)    # (frames, 53, 3)

    # a marker whose x, y AND z are all exactly 0 was lost -> NaN
    lost = np.all(arr == 0.0, axis=2)                     # (frames, 53)
    arr[lost] = np.nan
    return arr


def get_marker_names(txt_path) -> list:
    """Return the 53 face-marker names in order, e.g. ['CH1', ..., 'Mou8'].

    Useful later for building the graph (which markers are neighbours).
    """
    header = Path(txt_path).read_text(errors="ignore").splitlines()[0].split()
    names = header[2:]                    # drop 'Frame#' and 'Time'
    return names[:N_FACE_MARKERS]         # drop LHD, RHD
```
