```python
import shutil
from pathlib import Path
import kagglehub

# -----------------------------
# Dataset handles
# -----------------------------
RAW_DATASET = "dejolilandry/iemocapfullrelease"
CSV_DATASET = "samuelsamsudinng/iemocap-emotion-speech-database"

# -----------------------------
# Paths
# -----------------------------
RAW_DIR = DATA_DIR / "IEMOCAP_full_release"
CSV_DEST = RAW_DIR / "iemocap_full_dataset.csv"

# =====================================================
# Step 1: Download raw dataset if needed
# =====================================================
if not RAW_DIR.exists():
    print("Downloading IEMOCAP raw dataset...")

    raw_path = kagglehub.dataset_download(
        RAW_DATASET,
        output_dir=str(DATA_DIR)
    )

    print("Raw dataset downloaded.")
    print("Downloaded to:", raw_path)

else:
    print("Raw dataset already exists.")

# =====================================================
# Step 2: Copy CSV into raw dataset folder if needed
# =====================================================
if not CSV_DEST.exists():

    print("Downloading CSV dataset...")

    csv_cache = kagglehub.dataset_download(CSV_DATASET)

    csv_cache = Path(csv_cache)

    # Find the CSV file inside the downloaded dataset
    csv_source = next(csv_cache.rglob("*.csv"))

    # Copy it into the raw dataset folder
    shutil.copy2(csv_source, CSV_DEST)

    print("CSV copied successfully.")
    print("Source:", csv_source)
    print("Destination:", CSV_DEST)

else:
    print("CSV already exists.")

print("\nDataset location:", RAW_DIR)
```
